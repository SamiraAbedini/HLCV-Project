```python
# ============================================================
# DocTamper Dataset Full Inspection Notebook Cell
# ============================================================
# This code inspects the extracted DocTamper dataset.
# It does NOT load the full dataset into memory.
# It reads one sample at a time from the LMDB files.
# ============================================================

from pathlib import Path
import io
import random

import lmdb
import cv2
import numpy as np
import pandas as pd
from PIL import Image
import matplotlib.pyplot as plt
from IPython.display import display, Markdown


# ============================================================
# 1. Settings
# ============================================================
PROJECT_ROOT = Path.cwd()

possible_data_roots = [
    PROJECT_ROOT / "data" / "doctamper_data",
    PROJECT_ROOT.parent / "data" / "doctamper_data",
    PROJECT_ROOT.parent.parent / "data" / "doctamper_data",
]

DATA_ROOT = None

for candidate in possible_data_roots:
    candidate = candidate.resolve()
    if candidate.exists() and any(candidate.rglob("data.mdb")):
        DATA_ROOT = candidate
        break

if DATA_ROOT is None:
    raise FileNotFoundError(
        "Could not find DocTamper data. Please update DATA_ROOT manually."
    )

OUTPUT_DIR = DATA_ROOT.parent / "doctamper_analysis_outputs"
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

print("Using DATA_ROOT:", DATA_ROOT)
print("Using OUTPUT_DIR:", OUTPUT_DIR)

N_STATS_PER_SUBSET = 200            # random samples to analyze statistically from each subset. (increase to 1000 for more reliable stats)
N_VISUAL_SAMPLES_PER_SUBSET = 3     # visual examples to show from each subset
RANDOM_SEED = 42
random.seed(RANDOM_SEED)

print("=" * 80)
print("DocTamper Dataset Inspection")
print("=" * 80)
print(f"DATA_ROOT:  {DATA_ROOT}")
print(f"OUTPUT_DIR: {OUTPUT_DIR}")
print(f"N_STATS_PER_SUBSET: {N_STATS_PER_SUBSET}")
print(f"N_VISUAL_SAMPLES_PER_SUBSET: {N_VISUAL_SAMPLES_PER_SUBSET}")
print("=" * 80)


# ============================================================
# 2. Helper functions
# ============================================================

def find_lmdb_dirs(data_root):
    """
    Find all folders that contain data.mdb and lock.mdb.
    Each such folder is one DocTamper subset.
    """
    lmdb_dirs = []
    for p in data_root.rglob("*"):
        if p.is_dir() and (p / "data.mdb").exists() and (p / "lock.mdb").exists():
            lmdb_dirs.append(p)
    return sorted(lmdb_dirs)


def open_lmdb_env(lmdb_dir):
    """
    Open one LMDB folder in read-only mode.
    """
    return lmdb.open(
        str(lmdb_dir),
        readonly=True,
        lock=False,
        readahead=False,
        meminit=False
    )


def get_num_samples(lmdb_dir):
    """
    Read the number of samples stored in one LMDB subset.
    """
    env = open_lmdb_env(lmdb_dir)
    with env.begin(write=False) as txn:
        n = txn.get(b"num-samples")
    env.close()
    return int(n) if n is not None else None


def show_first_keys(lmdb_dir, max_keys=20):
    """
    Show the first few keys inside one LMDB file.
    This helps us understand the structure.
    """
    env = open_lmdb_env(lmdb_dir)
    keys = []

    with env.begin(write=False) as txn:
        cursor = txn.cursor()
        for i, (key, value) in enumerate(cursor):
            if i >= max_keys:
                break
            keys.append(key.decode("utf-8", errors="ignore"))

    env.close()
    return keys


def read_lmdb_sample(lmdb_dir, index):
    """
    Read one document image and its ground-truth tamper mask.
    image-000000000 = document image
    label-000000000 = mask
    """
    env = open_lmdb_env(lmdb_dir)

    with env.begin(write=False) as txn:
        img_key = f"image-{index:09d}".encode("utf-8")
        lbl_key = f"label-{index:09d}".encode("utf-8")

        imgbuf = txn.get(img_key)
        lblbuf = txn.get(lbl_key)

        if imgbuf is None:
            raise KeyError(f"Missing image key: {img_key}")
        if lblbuf is None:
            raise KeyError(f"Missing label key: {lbl_key}")

        image = Image.open(io.BytesIO(imgbuf)).convert("RGB")
        mask = cv2.imdecode(np.frombuffer(lblbuf, dtype=np.uint8), cv2.IMREAD_GRAYSCALE)

    env.close()
    return image, mask


def get_mask_bbox(binary_mask):
    """
    Get bounding box around tampered pixels.
    Returns None if the mask is empty.
    """
    ys, xs = np.where(binary_mask)

    if len(xs) == 0 or len(ys) == 0:
        return None

    x_min, x_max = int(xs.min()), int(xs.max())
    y_min, y_max = int(ys.min()), int(ys.max())

    return {
        "x_min": x_min,
        "y_min": y_min,
        "x_max": x_max,
        "y_max": y_max,
        "bbox_width": x_max - x_min + 1,
        "bbox_height": y_max - y_min + 1,
        "bbox_area": (x_max - x_min + 1) * (y_max - y_min + 1)
    }


def count_connected_components(binary_mask):
    """
    Count connected tampered regions in the mask.
    This gives a rough idea whether tampering is one region or multiple regions.
    """
    binary_uint8 = binary_mask.astype(np.uint8)
    num_labels, labels = cv2.connectedComponents(binary_uint8)

    # subtract 1 because label 0 is background
    return max(0, num_labels - 1)


def inspect_one_sample(lmdb_dir, index):
    """
    Read one sample and return useful statistics.
    """
    image, mask = read_lmdb_sample(lmdb_dir, index)

    width, height = image.size
    binary_mask = mask > 0

    tampered_pixels = int(binary_mask.sum())
    total_pixels = int(binary_mask.size)
    tampered_ratio = tampered_pixels / total_pixels

    bbox = get_mask_bbox(binary_mask)
    num_components = count_connected_components(binary_mask)

    row = {
        "subset": lmdb_dir.name,
        "index": index,
        "image_width": width,
        "image_height": height,
        "mask_width": mask.shape[1],
        "mask_height": mask.shape[0],
        "image_mask_same_size": (width == mask.shape[1] and height == mask.shape[0]),
        "mask_dtype": str(mask.dtype),
        "mask_min": int(mask.min()),
        "mask_max": int(mask.max()),
        "mask_unique_values": str(np.unique(mask)[:20].tolist()),
        "tampered_pixels": tampered_pixels,
        "total_pixels": total_pixels,
        "tampered_ratio": tampered_ratio,
        "tampered_ratio_percent": tampered_ratio * 100,
        "has_tamper_pixels": tampered_pixels > 0,
        "num_connected_tamper_components": num_components,
    }

    if bbox is None:
        row.update({
            "bbox_x_min": None,
            "bbox_y_min": None,
            "bbox_x_max": None,
            "bbox_y_max": None,
            "bbox_width": None,
            "bbox_height": None,
            "bbox_area": None,
            "bbox_area_ratio": None,
        })
    else:
        row.update({
            "bbox_x_min": bbox["x_min"],
            "bbox_y_min": bbox["y_min"],
            "bbox_x_max": bbox["x_max"],
            "bbox_y_max": bbox["y_max"],
            "bbox_width": bbox["bbox_width"],
            "bbox_height": bbox["bbox_height"],
            "bbox_area": bbox["bbox_area"],
            "bbox_area_ratio": bbox["bbox_area"] / total_pixels,
        })

    return row, image, mask


def make_overlay(image, mask, alpha=0.45):
    """
    Create an RGB overlay image.
    Tampered pixels are shown in red.
    """
    image_np = np.array(image).copy()
    binary_mask = mask > 0

    overlay = image_np.copy()
    overlay[binary_mask] = [255, 0, 0]

    blended = image_np.copy()
    blended[binary_mask] = (
        alpha * overlay[binary_mask] + (1 - alpha) * image_np[binary_mask]
    ).astype(np.uint8)

    return Image.fromarray(blended)


def display_sample(image, mask, title=""):
    """
    Display document image, mask, and overlay.
    """
    binary_mask = mask > 0
    overlay = make_overlay(image, mask)

    plt.figure(figsize=(8, 8))
    plt.imshow(image)
    plt.title(f"{title} - Document image")
    plt.axis("off")
    plt.show()

    plt.figure(figsize=(8, 8))
    plt.imshow(binary_mask, cmap="gray")
    plt.title(f"{title} - Ground-truth tamper mask")
    plt.axis("off")
    plt.show()

    plt.figure(figsize=(8, 8))
    plt.imshow(overlay)
    plt.title(f"{title} - Overlay, tampered pixels in red")
    plt.axis("off")
    plt.show()


def save_sample_outputs(image, mask, output_folder, prefix):
    """
    Save image, mask, and overlay to disk.
    """
    output_folder.mkdir(parents=True, exist_ok=True)

    binary_mask = (mask > 0).astype(np.uint8) * 255
    overlay = make_overlay(image, mask)

    image_path = output_folder / f"{prefix}_image.jpg"
    mask_path = output_folder / f"{prefix}_mask.png"
    overlay_path = output_folder / f"{prefix}_overlay.jpg"

    image.save(image_path)
    cv2.imwrite(str(mask_path), binary_mask)
    overlay.save(overlay_path)

    return image_path, mask_path, overlay_path


# ============================================================
# 3. Find dataset subsets
# ============================================================

print("\n" + "=" * 80)
print("STEP 1: Finding LMDB subset folders")
print("=" * 80)

lmdb_dirs = find_lmdb_dirs(DATA_ROOT)

if not lmdb_dirs:
    raise FileNotFoundError(
        f"No LMDB folders found under {DATA_ROOT}. "
        "Check that the dataset was extracted correctly."
    )

print("Found LMDB folders:")
for p in lmdb_dirs:
    print(" -", p)

print(f"\nNumber of subsets found: {len(lmdb_dirs)}")


# ============================================================
# 4. Create main dataset summary table
# ============================================================

print("\n" + "=" * 80)
print("STEP 2: Creating dataset summary table")
print("=" * 80)

summary_rows = []

for p in lmdb_dirs:
    n_samples = get_num_samples(p)
    data_size_gb = (p / "data.mdb").stat().st_size / (1024 ** 3)
    lock_size_kb = (p / "lock.mdb").stat().st_size / 1024

    summary_rows.append({
        "subset": p.name,
        "num_samples": n_samples,
        "data_mdb_size_GB": round(data_size_gb, 3),
        "lock_mdb_size_KB": round(lock_size_kb, 1),
        "path": str(p),
    })

summary_df = pd.DataFrame(summary_rows)
summary_df = summary_df.sort_values("num_samples").reset_index(drop=True)

print("Dataset summary:")
display(summary_df)

print("\nTotal number of samples:", int(summary_df["num_samples"].sum()))
print("Total data.mdb size in GB:", round(summary_df["data_mdb_size_GB"].sum(), 3))


# ============================================================
# 5. Inspect LMDB keys
# ============================================================

print("\n" + "=" * 80)
print("STEP 3: Inspecting LMDB keys")
print("=" * 80)

key_rows = []

for p in lmdb_dirs:
    keys = show_first_keys(p, max_keys=20)

    print(f"\nSubset: {p.name}")
    print("First keys:")
    for k in keys:
        print(" ", k)

    key_rows.append({
        "subset": p.name,
        "first_keys": ", ".join(keys[:10]),
        "contains_num_samples_key": "num-samples" in keys
    })

keys_df = pd.DataFrame(key_rows)
display(keys_df)


# ============================================================
# 6. Analyze random samples statistically
# ============================================================

print("\n" + "=" * 80)
print("STEP 4: Analyzing random samples from each subset")
print("=" * 80)

all_sample_rows = []
selected_visual_samples = {}

for p in lmdb_dirs:
    num_samples = get_num_samples(p)

    # Pick random samples for statistics
    n_to_check = min(N_STATS_PER_SUBSET, num_samples)
    stats_indices = random.sample(range(num_samples), n_to_check)

    # Pick random samples for visual display
    n_visual = min(N_VISUAL_SAMPLES_PER_SUBSET, num_samples)
    visual_indices = random.sample(range(num_samples), n_visual)
    selected_visual_samples[p.name] = visual_indices

    print(f"\nAnalyzing subset: {p.name}")
    print(f"Total samples: {num_samples}")
    print(f"Random samples used for statistics: {n_to_check}")
    print(f"Random samples selected for visualization: {visual_indices}")

    for idx in stats_indices:
        row, _, _ = inspect_one_sample(p, idx)
        all_sample_rows.append(row)

stats_df = pd.DataFrame(all_sample_rows)

print("\nFirst rows of sample-level statistics:")
display(stats_df.head())

# Save sample-level statistics
stats_csv_path = OUTPUT_DIR / "doctamper_sample_level_statistics.csv"
stats_df.to_csv(stats_csv_path, index=False)
print(f"\nSaved sample-level statistics to: {stats_csv_path}")


# ============================================================
# 7. Create subset-level statistics table
# ============================================================

print("\n" + "=" * 80)
print("STEP 5: Creating subset-level statistics table")
print("=" * 80)

subset_stats_df = stats_df.groupby("subset").agg(
    num_checked=("index", "count"),

    avg_image_width=("image_width", "mean"),
    min_image_width=("image_width", "min"),
    max_image_width=("image_width", "max"),

    avg_image_height=("image_height", "mean"),
    min_image_height=("image_height", "min"),
    max_image_height=("image_height", "max"),

    all_image_mask_same_size=("image_mask_same_size", "all"),

    avg_tampered_pixels=("tampered_pixels", "mean"),
    min_tampered_pixels=("tampered_pixels", "min"),
    max_tampered_pixels=("tampered_pixels", "max"),

    avg_tampered_ratio_percent=("tampered_ratio_percent", "mean"),
    min_tampered_ratio_percent=("tampered_ratio_percent", "min"),
    max_tampered_ratio_percent=("tampered_ratio_percent", "max"),

    percent_samples_with_tamper=("has_tamper_pixels", "mean"),

    avg_connected_components=("num_connected_tamper_components", "mean"),
    max_connected_components=("num_connected_tamper_components", "max"),

    avg_bbox_width=("bbox_width", "mean"),
    avg_bbox_height=("bbox_height", "mean"),
    avg_bbox_area_ratio=("bbox_area_ratio", "mean"),
).reset_index()

# Convert proportion to percentage
subset_stats_df["percent_samples_with_tamper"] *= 100

# Round for readability
round_cols = [
    "avg_image_width", "avg_image_height",
    "avg_tampered_pixels",
    "avg_tampered_ratio_percent",
    "min_tampered_ratio_percent",
    "max_tampered_ratio_percent",
    "percent_samples_with_tamper",
    "avg_connected_components",
    "avg_bbox_width",
    "avg_bbox_height",
    "avg_bbox_area_ratio",
]

for col in round_cols:
    if col in subset_stats_df.columns:
        subset_stats_df[col] = subset_stats_df[col].round(5)

print("Subset-level statistics based on random samples:")
display(subset_stats_df)

subset_stats_csv_path = OUTPUT_DIR / "doctamper_subset_statistics.csv"
subset_stats_df.to_csv(subset_stats_csv_path, index=False)
print(f"\nSaved subset-level statistics to: {subset_stats_csv_path}")


# ============================================================
# 8. Print clear interpretation 
# ============================================================

print("\n" + "=" * 80)
print("STEP 6: Simple interpretation")
print("=" * 80)

print("""
What we learned so far:

1. The dataset is stored in LMDB format, not normal image folders.
2. Each subset folder contains:
   - data.mdb: the actual image/mask database
   - lock.mdb: a small LMDB helper file
3. Inside each LMDB, the keys follow this pattern:
   - image-000000000 = document image
   - label-000000000 = ground-truth tamper mask
   - num-samples = number of samples
4. The label/mask is pixel-level.
5. Pixels with value > 0 are treated as tampered pixels.
6. Tampered regions are usually very small compared to the full image.
""")


# ============================================================
# 9. Display and save random visual samples
# ============================================================

print("\n" + "=" * 80)
print("STEP 7: Displaying and saving random visual samples")
print("=" * 80)

visual_report_rows = []

for p in lmdb_dirs:
    subset_output_dir = OUTPUT_DIR / "visual_examples" / p.name
    subset_output_dir.mkdir(parents=True, exist_ok=True)

    print("\n" + "-" * 80)
    print(f"Visual examples from subset: {p.name}")
    print("-" * 80)

    for idx in selected_visual_samples[p.name]:
        row, image, mask = inspect_one_sample(p, idx)

        print(f"\nSubset: {p.name}, Index: {idx}")
        print(f"Image size: {image.size}")
        print(f"Mask shape: {mask.shape}")
        print(f"Mask unique values: {np.unique(mask)[:20]}")
        print(f"Tampered pixels: {row['tampered_pixels']}")
        print(f"Tampered ratio: {row['tampered_ratio_percent']:.6f}%")
        print(f"Connected tamper components: {row['num_connected_tamper_components']}")

        if row["bbox_width"] is not None:
            print(
                f"Tamper bounding box: "
                f"x={row['bbox_x_min']}..{row['bbox_x_max']}, "
                f"y={row['bbox_y_min']}..{row['bbox_y_max']}, "
                f"width={row['bbox_width']}, height={row['bbox_height']}"
            )
        else:
            print("Tamper bounding box: None")

        prefix = f"{p.name}_{idx:09d}"
        image_path, mask_path, overlay_path = save_sample_outputs(
            image=image,
            mask=mask,
            output_folder=subset_output_dir,
            prefix=prefix
        )

        visual_report_rows.append({
            "subset": p.name,
            "index": idx,
            "image_path": str(image_path),
            "mask_path": str(mask_path),
            "overlay_path": str(overlay_path),
            "tampered_ratio_percent": row["tampered_ratio_percent"],
            "tampered_pixels": row["tampered_pixels"],
            "num_connected_components": row["num_connected_tamper_components"],
        })

        display_sample(image, mask, title=f"{p.name}, index {idx}")

visual_examples_df = pd.DataFrame(visual_report_rows)
visual_examples_csv_path = OUTPUT_DIR / "doctamper_visual_examples.csv"
visual_examples_df.to_csv(visual_examples_csv_path, index=False)

print(f"\nSaved visual example metadata to: {visual_examples_csv_path}")
print(f"Saved visual example images to: {OUTPUT_DIR / 'visual_examples'}")
display(visual_examples_df)


# ============================================================
# 10. Plot simple statistics
# ============================================================

print("\n" + "=" * 80)
print("STEP 8: Plotting simple statistics")
print("=" * 80)

# Plot number of samples per subset
plt.figure(figsize=(10, 5))
plt.bar(summary_df["subset"], summary_df["num_samples"])
plt.xticks(rotation=30, ha="right")
plt.ylabel("Number of samples")
plt.title("DocTamper samples per subset")
plt.tight_layout()
plt.show()

# Plot data.mdb size per subset
plt.figure(figsize=(10, 5))
plt.bar(summary_df["subset"], summary_df["data_mdb_size_GB"])
plt.xticks(rotation=30, ha="right")
plt.ylabel("data.mdb size in GB")
plt.title("DocTamper LMDB size per subset")
plt.tight_layout()
plt.show()

# Plot average tampered ratio per subset
plt.figure(figsize=(10, 5))
plt.bar(subset_stats_df["subset"], subset_stats_df["avg_tampered_ratio_percent"])
plt.xticks(rotation=30, ha="right")
plt.ylabel("Average tampered pixels (%)")
plt.title(f"Average tampered pixel ratio per subset, based on {N_STATS_PER_SUBSET} samples/subset")
plt.tight_layout()
plt.show()

# Plot distribution of tampered ratios
plt.figure(figsize=(10, 5))
plt.hist(stats_df["tampered_ratio_percent"], bins=40)
plt.xlabel("Tampered pixel ratio (%)")
plt.ylabel("Number of checked samples")
plt.title("Distribution of tampered pixel ratios in random sample")
plt.tight_layout()
plt.show()


# ============================================================
# 11. Save a Markdown report
# ============================================================

print("\n" + "=" * 80)
print("STEP 9: Saving Markdown report")
print("=" * 80)

report_path = OUTPUT_DIR / "doctamper_dataset_report.md"

with open(report_path, "w") as f:
    f.write("# DocTamper Dataset Inspection Report\n\n")

    f.write("## Dataset location\n\n")
    f.write(f"- Extracted data root: `{DATA_ROOT}`\n")
    f.write(f"- Analysis output folder: `{OUTPUT_DIR}`\n\n")

    f.write("## Dataset structure\n\n")
    f.write("The dataset is stored as LMDB databases. Each subset contains:\n\n")
    f.write("- `data.mdb`: database containing document images and masks\n")
    f.write("- `lock.mdb`: small LMDB lock/helper file\n\n")

    f.write("## Subset summary\n\n")
    f.write(summary_df.to_markdown(index=False))
    f.write("\n\n")

    f.write("## LMDB key format\n\n")
    f.write("Each LMDB contains keys like:\n\n")
    f.write("- `image-000000000`: document image\n")
    f.write("- `label-000000000`: corresponding ground-truth tamper mask\n")
    f.write("- `num-samples`: number of samples in the subset\n\n")

    f.write("## Mask interpretation\n\n")
    f.write("The label is a pixel-level ground-truth tampering mask. ")
    f.write("Pixels with value greater than 0 are treated as tampered pixels. ")
    f.write("The tampered regions are usually very small compared to the full document image.\n\n")

    f.write("## Sample-based subset statistics\n\n")
    f.write(
        f"The following statistics are calculated from "
        f"{N_STATS_PER_SUBSET} random samples per subset.\n\n"
    )
    f.write(subset_stats_df.to_markdown(index=False))
    f.write("\n\n")

    f.write("## Visual examples\n\n")
    f.write("Random visual examples were saved under:\n\n")
    f.write(f"`{OUTPUT_DIR / 'visual_examples'}`\n\n")
    f.write(visual_examples_df.to_markdown(index=False))
    f.write("\n")

print(f"Markdown report saved to: {report_path}")

display(Markdown(f"""
## Finished

The DocTamper dataset inspection is complete.

Generated files:

- Summary statistics CSV: `{subset_stats_csv_path}`
- Sample-level statistics CSV: `{stats_csv_path}`
- Visual examples CSV: `{visual_examples_csv_path}`
- Markdown report: `{report_path}`
- Saved images/masks/overlays: `{OUTPUT_DIR / "visual_examples"}`
"""))
```
